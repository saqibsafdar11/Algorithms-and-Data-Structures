```
# adj is a list of lists, where adj[i] is the list of neighbors of vertex i.
# M is the adjacency matrix of the graph.
# n is the number of vertices in the graph.

# Step 1: For each vertex v, look at every pair of its neighbors (call them x and y).
# Step 2- If x and y are not directly connected (i.e. there is no edge between them), then record that v is a common neighbor for the pair (x, y).
# Step 3: For each pair (x, y) that has at least two common neighbors, examine all pairs of these common neighbors.
# Step 4: If any two common neighbors (say a and b) are not connected, then the vertices x, a, y, and b (or a similar ordering) form a cycle, so return True. If no such cycle is found after all checks, return False.

function cycle(n, adj, M):
    common = empty dictionary

    for v from 0 to n-1:
        neighbors = sorted(adj[v])
        for i from 0 to length(neighbors) - 1:
            for j from i + 1 to length(neighbors) - 1:
                x = neighbors[i]
                y = neighbors[j]
                if M[x][y] == 0:
                    if (x, y) not in common:
                        common[(x, y)] = empty list
                    append v to common[(x, y)]

    for each key (x, y) in common:
        commonList = common[(x, y)]
        if length(commonList) < 2:
            continue
        for i from 0 to length(commonList) - 1:
            for j from i + 1 to length(commonList) - 1:
                a = commonList[i]
                b = commonList[j]
                if M[a][b] == 0:
                    return True

    return False
```

#### Data Structure and Purpose
- adjacency list: to store the graph in a way that allows us to easily access the neighbors of a vertex
- adjacency matrix: to store the graph in a way that allows us to easily check if two vertices are connected

#### Running Time Analysis
- m is the number of edges, n is the number of vertices
- adjacency list: O(m+n)
- adjacency matrix: O(n^2)
- for all pairs of vertices, we can check if they are connected in O(1) time using adjacency matrix 
- to check neightbors of a vertex, we can do it in O(deg(v)) time using adjacency list
- all pairs of neighbors of a vertex can be checked in O(deg(v)^2) time 
- over all vertices, we get sum of deg(v)^2
- in a dense graph, m = O(n^2), so sum of deg(v)^2
- in a sparse graph, m = O(n), so sum of deg(v)^2 = O(n)
- each time we ask if x-y is connected, it takes O(1) time
- combinbing the results O(m^2 + n^2) in every possible case
